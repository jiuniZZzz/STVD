# SimVP 训练示例
使用 STVD/SeaFog 的 .npy 序列数据训练 SimVP 并保存参数。

```python
import torch
from torch.utils.data import DataLoader

from data.seafog_dataset import SeaFogSTVDDataset
from model.simvp import SimVP
from trainer_utils import train_forecast_model

npy_path = "/path/to/your/dataset.npy"  # 修改为你的数据集路径
checkpoint_path = "simvp_checkpoint.pt"

dataset = SeaFogSTVDDataset(npy_path, input_frames=4)
dataloader = DataLoader(dataset, batch_size=4, shuffle=True, num_workers=2)
model = SimVP(in_channels=3, hid_channels=64)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

model, history = train_forecast_model(
    model=model,
    dataloader=dataloader,
    epochs=10,
    optimizer=optimizer,
    save_path=checkpoint_path,
)
```
